### Test Case 1

### Test Case 4
- Objective: To verify whether the model can correctly recognize apple images under low light conditions
- Input: `images/apple1.png`, `images/apple2.png`, `images/apple3.png`, The brightness has decreased to 40% of its original level
- Expected Result: The model returns' apple 'for all three low light images

```python
from PIL import Image, ImageEnhance
from model.model_inference import predict_image

for i in range(1, 4):
    image = Image.open(f"images/apple{i}.png")
    image_dark = ImageEnhance.Brightness(image).enhance(0.4)
    image_dark.save(f"images/apple{i}_dark.png")

    result = predict_image(f"images/apple{i}_dark.png")

    print(f"Apple{i} low-light prediction result:", result["Status"])

    if result["Status"] == "apple":
        print(f"Apple{i}: Test passed")
        print()
    else:
        print(f"Apple{i}: Test failed. \n", result)
        print()
```

### Test Case 5
- Objective: Verify whether the model returns data that meets the requirements
- Input: `images/apple1.png`
- Expected Result: The model returns a dictionary containing four fields: Status, Top Probability, Energy Score, and Detailed Probability

```python
from model.model_inference import predict_image

result = predict_image("images/apple1.png")

print("Model prediction result: ", result)

if (isinstance(result, dict)
    and "Status" in result
    and "Top Probability" in result
    and "Energy Score" in result
    and "Detailed Probabilities" in result):
    print("Test case 5 passed")
else:
    print("Test case 5 failed")
```

### Test Case 6
- Objective: Verify whether the predicted probability output by the model meets the requirements
- Input: `images/apple1.png`
- Expected Result: The predicted probabilities for Apple and Banana are between 0% and 100%

```python
from model.model_inference import predict_image

result = predict_image("images/apple1.png")

apple_prob = float(result["Detailed Probabilities"]["apple"].strip("%"))
banana_prob = float(result["Detailed Probabilities"]["banana"].strip("%"))

print("Apple probability:", apple_prob)
print("Banana probability:", banana_prob)

if 0 <= apple_prob <= 100 and 0 <= banana_prob <= 100:
    print("Test case 6 passed")
else:
    print("Test case 6 failed")
```
