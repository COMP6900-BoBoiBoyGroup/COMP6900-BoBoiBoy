### Test Case 1
- Objective: To verify whether the model can correctly recognize normal apple images.
- Input: `images/apple1.png`
- Expected Result: The model returns apple.

```sql
from backend.main import predict_image_wrapper

for i in range(1, 4):
    result = predict_image_wrapper(f"images/apple{i}.png")

    print(f"Apple{i} prediction result:", result["label"])

    if result["label"] == "apple":
        print(f"Apple{i}: Test passed")
        print()
    else:
        print(f"Apple{i}: Test failed.\n", result)
        print()
```

### Test Case 3
- Objective: Verify whether the backend returns data that meets the format required by the frontend
- Input: `images/apple1.png`
- Expected Result: The model returns a dictionary containing four fields: label, confidence, ripe, ripe_confidence

```sql
from backend.main import predict_image_wrapper

result = predict_image_wrapper("images/apple1.png")

print("Model prediction result: ", result)

if (isinstance(result, dict)
    and "label" in result
    and "confidence" in result
    and "ripe" in result
    and "ripe_confidence" in result):
    print("Test case 5 passed")
else:
    print("Test case 5 failed")
```

### Test Case 4
- Objective: Verify whether the predicted probability output by the model are properly parsed into a format the frontend accepts
- Input: `images/banana1.png`
- Expected Result: The predicted probabilities for Apple and Banana are converted to between 0 and 1

```sql
from backend.main import predict_image_wrapper

result = predict_image_wrapper("images/banana1.png")

banana_prob = float(result["confidence"])
banana_ripe_prob = float(result["ripe_confidence"])

print("Banana confidence:", banana_prob)
print("Banana ripe confidence:", banana_ripe_prob)

if 0 <= banana_prob <= 1:
    print("Test case 6 confidence: passed")
else:
    print("Test case 6 confidence: failed")
if 0 <= banana_ripe_prob <= 1:
    print("Test case 6 ripe confidence: passed")
else:
    print("Test case 6 ripe confidence: failed")
```

### Test Case 5

- Input:` images/apple1.png `, `images/unripe.webp `, `images/ripe.jpeg `, ` images/overripe.jpeg`, ` images/rotten.jpeg`
- Expected result: The backend correctly maps the result from the model to the correct value for the frontend
